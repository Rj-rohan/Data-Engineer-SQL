### Q26 Departments with salary above company average

```sql
SELECT Department,COUNT(*) FROM employees WHERE Salary > (SELECT AVG(Salary) FROM employees) GROUP BY Department;
```

### Q27 Count employees earning above department average

```sql
SELECT e1.Department, COUNT(*)
FROM employees e1
WHERE Salary >
(
    SELECT AVG(Salary)
    FROM employees e2
    WHERE e1.Department = e2.Department
)
GROUP BY e1.Department;
```

### Q28 Department having highest total salary

```sql
SELECT Department,SUm(Salary) FROM employees GROUP BY Department ORDER BY SUM(Salary) DESC LIMIT 1;
```

### Q29 Departments with at least 3 distinct salary values

```sql
SELECT Department,COUNT(DISTINCT salary) FROM employees GROUP BY Department HAVING COUNT(Distinct salary) >=3
```

## Duplicate Detection

### Find duplicate names

```sql
SELECT Name,COUNT(*) FROM employees GROUP BY Name HAVING COUNT(*)>1;
-- No Output bcz no duplicate
```

### Find duplicate salary values

```sql
SELECT Salary,COUNT(*) FROM employees GROUP BY Salary HAVING COUNT(*)>1;
```

### Find duplicate employees based on Name + Department

```sql
SELECT Name,Department,COUNT(*) FROM employees GROUP BY Name,Department HAVING COUNT(*)>1;
-- No Output bcz no duplicate
```

### Find duplicate employees based on Name + City + Department

```sql
SELECT Name, City, Department, COUNT(*)
FROM employees
GROUP BY Name, City, Department
HAVING COUNT(*) > 1;
```

### Find employees having same salary in same department

```sql
SELECT Department, Salary, COUNT(*)
FROM employees
GROUP BY Department, Salary
HAVING COUNT(*) > 1;
```

### Find departments having duplicate salaries

```sql
SELECT Department,COUNT(*) FROM employees GROUP BY Department,Salary HAVING COUNT(*)>1;
```

### Display actual duplicate records

```sql
SELECT * FROM employees WHERE Name IN(SELECT Name FROM employees GROUP BY Name HAVING COUNT(*)>1);
```

### Find duplicate records excluding EmpID

```sql
SELECT Name,
       Department,
       City,
       Gender,
       Salary,
       Experience,
       Project,
       COUNT(*)
FROM employees
GROUP BY Name,
         Department,
         City,
         Gender,
         Salary,
         Experience,
         Project
HAVING COUNT(*) > 1;
```

### Top 3 departments by average salary

```sql
SELECT Department,AVG(Salary) FROM employees GROUP BY Department ORDER BY AVG(Salary) DESC LIMIT 3;
```

### Bottom 2 cities by total salary

```sql
SELECT CITY,SUM(Salary) FROM employees GROUP BY City ORDER BY SUM(Salary) ASC LIMIT 2;
```

```sql
SELECT Project,min(Salary) FROm employees GROUP BY Project ORDER BY min(Salary) LIMIT 1;
```

### Find second highest average salary department


```sql
SELECT Department,AVG(Salary) FROM employees GROUP BY Department ORDER BY AVG(Salary) DESC LIMIT 1 OFFSET 1;
```
