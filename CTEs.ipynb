## CTE - Common table expression

```sql
with avg AS (select department,avg(salary) from window GROUP BY department)
select * from avg;

-- Cleaner and Professional , readable
```

```sql
-- WHY DATA ENGINEERS LOVE CTEs

-- Because ETL pipelines are step-by-step transformations.

-- Example:

-- raw data
-- cleaned data
-- transformed data
-- aggregated data
-- final reporting table

-- CTE matches this perfectly.
```

### Multiple CTEs

```sql
-- Goal

-- Find:

-- total spending
-- average spending
-- customers spending above average

 with customer_total AS (select sum(salary) AS total_spent from window GROUP BY emp_id),
 avg_total AS (select avg(salary) AS avg from window )

 select c.total_spent,a.avg from customer_total c,
             avg_total a
        WHERE c.total_spent > a.avg


```

## RECURSIVE CTE

```sql
-- Recursive CTE to generate a sequence of salary bands starting from the minimum salary

WITH RECURSIVE salary_band AS (
  SELECT MIN(salary) AS band_start, MIN(salary) + 1000 AS band_end
  FROM window
  UNION ALL
  SELECT band_end AS band_start, band_end + 1000 AS band_end
  FROM salary_band
  WHERE band_end < (SELECT MAX(salary) FROM window)
)
SELECT * FROM salary_band;
```
