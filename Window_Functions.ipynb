### OVER() Clause - Mandatory to use in Window function

```sql
SELECT name,salary,AVG(salary) OVER() AS avg from window;
```

### PARTITION BY

#### Department-wise average salary

```sql
SELECT name,department,salary,AVG(salary) OVER(PARTITION BY department) AS dept_avg from window; 
-- PARTITION BY is used to group the data by a specific column
-- Try with PARTITON BY with joining_date,gender,city,state,country
```

## ORDER BY

```sql
SELECT name,salary,SUM(salary) OVER(PARTITION BY department ORDER BY salary) AS dept_running_total from window;
```

## ROW_NUMBER()

```sql
-- Assigns unique sequence number.
-- Mandatory field is ORDER BY
SELECT name,department,salary,ROW_NUMBER() OVER(ORDER BY salary ) AS rn from window;
```

```sql
-- Assigns unique sequence number.
SELECT name,department,salary,ROW_NUMBER() OVER(PARTITION BY department ORDER BY salary ) AS rn from window;
```

## RANK()

```sql
-- Same rank for ties, skips numbers.
-- ORDER BY is mandatory field in RANK
select name,department,salary,RANK() OVER(ORDER BY salary) AS rank from window;
-- WHEN to use ? --- 1)Sports Leaderboard - If two players score same points:
-- 2) Company Salary Ranking - Employees with same salary should have same rank.
-- 3) Exam results
```

## DENSE_RANK()


```sql
-- Does not skip rank numbers
-- ORDER BY is mandatory field in RANK
select name,department,salary,DENSE_RANK() OVER(ORDER BY salary) AS dense_rank from window;
```

## UNBOUNDED PRECEDING

```sql
-- It defines where the window/frame starts.
SELECT
    name,
    salary,
    SUM(salary) OVER(
        ORDER BY salary
    ) AS running_total
FROM window;
```

```sql
-- But HOW is SQL calculating this?

-- Internally SQL uses:

-- ROWS BETWEEN UNBOUNDED PRECEDING
-- AND CURRENT ROW

-- UNBOUNDED PRECEDING means Start from FIRST row of partition up to current row
-- Why We Use It

-- Used for:

-- Running totals
-- Cumulative sums
-- Cumulative averages
-- Progressive calculations
-- Financial reports
-- Trend analysis
```

### e.g 1) Running Total

```sql
SELECT
    name,
    salary,
    SUM(salary) OVER(
        ORDER BY salary
        ROWS BETWEEN UNBOUNDED PRECEDING
        AND CURRENT ROW
    ) AS running_total
FROM window;
```

### e.g 2 ) Running Avg

```sql
SELECT
    name,
    salary,
    AVG(salary) OVER(
        ORDER BY salary
        ROWS BETWEEN UNBOUNDED PRECEDING
        AND CURRENT ROW
    ) AS running_avg
FROM window;
```

### Take: 2 previous rows + current row

```sql
SELECT
    name,
    salary,
    SUM(salary) OVER(
        ORDER BY salary
        ROWS BETWEEN 2 PRECEDING
        AND CURRENT ROW
    ) AS moving_sum
FROM window;
```

### Full Partition - ROWS BETWEEN UNBOUNDED PRECEDING AND UNBOUNDED FOLLOWING
### Future Rows - ROWS BETWEEN CURRENT ROW AND UNBOUNDED FOLLOWING - Current row → last row
