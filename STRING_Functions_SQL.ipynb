## STRING Functions

## UPPER()

#### Converts to uppercase.

```sql
SELECT UPPER(name) FROM window;
```

## LOWER()
#### Converts to Lowercase

```sql
SELECT LOWER(name) FROM window;
```

## LENGTH()
#### Calculating length of name

```sql
SELECT name,LENGTH(name) FROM window;
```

## CONCAT()
#### Combine name and city

```sql
SELECT CONCAT(name,' lives in ',city)
FROM window;
```

## SUBSTRING()
#### First 3 characters of names

```sql
SELECT name,SUBSTRING(name,1,3) FROM window;
```

## REPLACE()
Replace Pune with Pune City

```sql
SELECT city,REPLACE(city,'Pune','Pune City') from window;
```

## TRIM()

#### Removes spaces.

```sql
select name,TRIM(name) FROM window;
```
