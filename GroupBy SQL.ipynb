## Level 1: Basic GROUP BY

### Q1 Count employees in each department



```sql
SELECT Department,COUNT(*) FROM employees GROUP BY Department;
```

### Q2 Find total salary department-wise

```sql
SELECT Department,SUM(Salary) FROM employees GROUP BY Department;
```

### Q3 Find average salary department-wise

```sql
SELECT Department,AVG(Salary) AS avg FROM employees GROUP BY Department;
```

### Q4 Find maximum salary in each department


```sql
SELECT Department,MAX(Salary) AS max_Salary FROM employees GROUP BY Department;
```

### Q5 Find minimum salary in each department

```sql
SELECT Department,MIN(Salary) FROM employees GROUP BY Department;
```
