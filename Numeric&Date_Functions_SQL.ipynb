## Numeric Functions

### ROUND()
#### Round average salary

```sql
SELECT department,ROUND(AVG(SALARY)) FROM window GROUP BY department;
```

###CEIL()
#### Rounds upward.

```sql
SELECT department,CEIL(AVG(SALARY)) FROM window GROUP BY department;
```

### FLOOR()

#### Rounds downward.

```sql
SELECT department,FLOOR(AVG(SALARY)) FROM window GROUP BY department;
```

### MOD()

#### Remainder - Check even/odd emp_id

```sql
SELECT emp_id,MOD(emp_id,2) FROM window;
```
